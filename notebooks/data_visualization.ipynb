**1- Importing the Required Libraries**

We import the libraries needed for data analysis, machine learning, visualization, and statistical plotting: Pandas, NumPy, Scikit-learn, Matplotlib, and Seaborn.

```python
import sys
import os

import pandas as pd
import numpy as np
import sklearn
import matplotlib.pyplot as plt
import seaborn as sns 
pd.set_option('display.float_format', lambda x: '%.2f' % x)
```

```python

sys.path.append(os.path.abspath('..'))

from config.path import DATA_RAW_PATH, DATA_PROCESSED_PATH
```

```python
voiture_df = pd.read_csv(DATA_RAW_PATH)
```

```python
plt.figure(figsize=(10,6))


plt.hist(voiture_df['selling_price']/ 1e6, bins=50, color='skyblue', edgecolor='black')

plt.xlim(0, 3)

plt.title('Distribution of Selling Price', fontsize=16)
plt.xlabel('Selling Price (in millions)', fontsize=14)
plt.ylabel('Frequency', fontsize=14)



plt.grid(axis='y', alpha=0.50)

plt.show()

```

```python
plt.figure(figsize=(10,6))

plt.title('Distribution of Fuel Types', fontsize=16)
plt.xlabel('Fuel Type', fontsize=14)
plt.ylabel('Count', fontsize=14)

fuel_counts = voiture_df['fuel'].value_counts()
plt.bar(fuel_counts.index, fuel_counts.values, color='skyblue', edgecolor='black')
```

```python
plt.figure(figsize=(10,6))

plt.title('Distribution of Seller Types', fontsize=16)  
plt.xlabel('Seller Type', fontsize=14)
plt.ylabel('Count', fontsize=14)

seller_type = voiture_df['seller_type'].value_counts()
plt.bar(seller_type.index, seller_type.values, color='skyblue', edgecolor='black')
```

```python
plt.figure(figsize=(10,6))

plt.hexbin(voiture_df['selling_price'] / 1e6, voiture_df['km_driven'], gridsize=50, cmap='Blues', mincnt=1)
plt.colorbar(label='Count of Cars') 

plt.xlim(0, 2)
plt.ylim(0, 400000)
plt.title('Car Density: Price vs. KM Driven', fontsize=16)
plt.xlabel('Selling Price (in millions)', fontsize=14)
plt.ylabel('Kilometers Driven', fontsize=14)
plt.show()
```

```python
avg_price_by_year = voiture_df.groupby('year')['selling_price'].mean()

plt.figure(figsize=(12, 6))

plt.plot(avg_price_by_year.index, avg_price_by_year.values / 1e6, marker='o', color='skyblue')

plt.title('Average Selling Price by Year', fontsize=16)
plt.xlabel('Year', fontsize=14)
plt.ylabel('Average Selling Price (in millions)', fontsize=14)


plt.grid(axis='y', alpha=0.50)
plt.show()
```

```python
corr = voiture_df.select_dtypes(include=[np.number]).corr()

plt.figure(figsize=(12, 8))
plt.imshow(corr, cmap='magma', interpolation='none')
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=90)
plt.yticks(range(len(corr)), corr.columns)
plt.title('Correlation Matrix', fontsize=16)

plt.show()
```

```python
nan_counts = voiture_df.isnull().sum()
nan_counts = nan_counts[nan_counts > 0]

plt.figure(figsize=(8, 4))
plt.bar(nan_counts.index, nan_counts.values, color='crimson')
plt.title('Count of Missing Values (NaN) per Column')
plt.xlabel('Columns')
plt.ylabel('Missing Values Count')
plt.show()
```

```python

data = voiture_df['km_driven'].dropna()

mean_val = data.mean()
median_val = data.median()
mode_val = data.mode()[0]

plt.figure(figsize=(10, 5))
plt.hist(data, bins=40, color='lightblue', edgecolor='black', alpha=0.7)

plt.xlim(0, 4e5)

plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.1f}')
plt.axvline(median_val, color='green', linestyle='-', linewidth=2, label=f'Median: {median_val:.1f}')
plt.axvline(mode_val, color='orange', linestyle=':', linewidth=2, label=f'Mode: {mode_val:.1f}')

plt.title('Distribution of KM Driven with Mean, Median, and Mode')
plt.xlabel('KM Driven')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 3))
plt.boxplot(data, vert=False)
plt.title('Boxplot of KM Driven (Outliers Detection)')
plt.xlabel('KM Driven')
plt.show()
```
