**1- Importing the Required Libraries**

We import the libraries needed for data analysis, machine learning, visualization, and statistical plotting: Pandas, NumPy, Scikit-learn, Matplotlib, and Seaborn.

```python
import sys
import os

import pandas as pd
import numpy as np
import sklearn
import matplotlib.pyplot as plt
import seaborn as sns 
pd.set_option('display.float_format', lambda x: '%.2f' % x)
```

```python

sys.path.append(os.path.abspath('..'))

from config.path import DATA_RAW_PATH
```

```python
df = pd.read_csv(DATA_RAW_PATH)
df_md = df.copy()
df_md.head()
```

```python
print(df_md.shape)
print(df_md.duplicated().sum())
```

```python
cat_col = ['name', 'fuel' , 'seller_type', 'transmission', 'owner']
num_col = ['year', 'selling_price', 'km_driven']
```

```python


for col in num_col : 
    df_md[col] =  df_md[col].fillna(df_md[col].median())

for col in num_col :
    print(f"Missing values in {col}: {df_md[col].isnull().sum()}")
```

```python
for col in cat_col : 
    if col == 'owner' : 
        continue 
    df_md[col] = df_md[col].fillna(df_md[col].mode()[0])
```

```python
owner_mapping = {
    'First Owner': 1,
    'Second Owner': 2,
    'Third Owner': 3,
    'Fourth & Above Owner': 4,
    'Test Drive Car': 5,
    np.nan: 0 
}

df_md['owner'] = df_md['owner'].map(owner_mapping)
```

```python
df_md = pd.get_dummies(df_md, columns=['fuel', 'transmission', 'seller_type'], dummy_na=True)
```

```python
df_md['brand'] = df_md['name'].apply(lambda x: x.split()[0])
df_md['brand'] = df_md['brand'].astype('category').cat.codes  
```

```python
df_md = df_md.drop(columns=['name'])
```

```python
df_md.to_csv('../data/processed/voiture_data_cleaned.csv', index=False)
```
